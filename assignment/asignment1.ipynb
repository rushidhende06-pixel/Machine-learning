```python
plt.figure(figsize=(6, 4))

sns.histplot(df['target'], kde=True)

plt.title("Distribution of Diabetes Target")
plt.xlabel("Target")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(6,4))

corr = df.corr()

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
#Find the most correlated feature
correlation_with_target = df.corr()['target'].drop('target').abs().sort_values(ascending=False)

print("Features ranked by absolute correlation with target:")
print(correlation_with_target)
```

```python
# Get most correlated feature
most_correlated_feature = correlation_with_target.index[0]

print("Most correlated feature:", most_correlated_feature)

# Scatter plot
plt.figure(figsize=(6, 4))

sns.scatterplot(
    x=df[most_correlated_feature],
    y=df['target']
)

plt.title(f"{most_correlated_feature} vs Target")
plt.xlabel(most_correlated_feature)
plt.ylabel("Target")
plt.show()
```

    *obervation:The scatter plot shows the relationship between the most correlated feature and the target variable. The direction of the relationship can be understood from the pattern of the points. A stronger correlation generally results in points following a clearer trend.*

Q4. Data Preprocessing

```python
X = df.drop('target', axis=1)
y = df['target']

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

Q5. Model Training

```python
from sklearn.linear_model import LinearRegression

# Create model
model = LinearRegression()

# Train model
model.fit(X_train, y_train)

# Coefficients
print("Model Coefficients:")
print(model.coef_)

# Intercept
print("\nModel Intercept:")
print(model.intercept_)
```

**Better way to display coefficients**

```python
coefficients = pd.DataFrame({
    'Feature': X.columns,
    'Coefficient': model.coef_
})

display(coefficients)
```

Q6. Prediction

```python
y_pred = model.predict(X_test)

print("Predicted Values:")
print(y_pred[:10])
```

```python
comparison = pd.DataFrame({
    'Actual': y_test.values,
    'Predicted': y_pred
})

display(comparison.head(20))
```

```python
comparison = pd.DataFrame({
    'Actual': y_test.values,
    'Predicted': y_pred
})

comparison['Predicted'] = comparison['Predicted'].round(2)

display(comparison.head(20))
```

    Q7. Model Evaluation

    Evaluate the model using:

    Mean Absolute Error (MAE)
    Mean Squared Error (MSE)
    Root Mean Squared Error (RMSE)
    R² Score

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
import numpy as np

# MAE
mae = mean_absolute_error(y_test, y_pred)

# MSE
mse = mean_squared_error(y_test, y_pred)

# RMSE
rmse = np.sqrt(mse)

# R2 Score
r2 = r2_score(y_test, y_pred)

print("Mean Absolute Error (MAE):", mae)
print("Mean Squared Error (MSE):", mse)
print("Root Mean Squared Error (RMSE):", rmse)
print("R² Score:", r2)
```

```python
coef_df = pd.DataFrame({
    'Feature': X.columns,
    'Coefficient': model.coef_
})

coef_df['Absolute_Coefficient'] = coef_df['Coefficient'].abs()

coef_df = coef_df.sort_values(
    by='Absolute_Coefficient',
    ascending=False
)

display(coef_df)
```

```python
most_influential = coef_df.iloc[0]

print("Most Influential Feature:", most_influential['Feature'])
print("Coefficient:", most_influential['Coefficient'])
```

```python
feature = most_influential['Feature']
coefficient = most_influential['Coefficient']

if coefficient > 0:
    print(f"{feature} has a positive effect on the target variable.")
else:
    print(f"{feature} has a negative effect on the target variable.")
```
